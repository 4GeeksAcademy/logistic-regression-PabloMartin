### Análisis de variables numéricas

- age: Edad del cliente (numérico)

- duration: Duración del contacto previo en segundos 

- campaign: Número de contactos realizados durante esta campaña al cliente 

- pdays: Número de días que transcurrieron desde la última campaña hasta que fue contactado 

- previous: Número de contactos realizados durante la campaña anterior al cliente 

- emp.var.rate: Tasa de variación del empleo. Indicador trimestral 

- cons.price.idx: Índice de precios al consumidor. Indicador mensual

- cons.conf.idx: Índice de confianza del consumidor. Indicador mensual 

- euribor3m: Tasa EURIBOR 3 meses. Indicador diario

- nr.employed: Número de empleados. Indicador trimestral 

```python
# Lista de variables numéricas
numericas = ['age', 'duration', 'campaign', 'pdays', 'previous', 'emp.var.rate', 'cons.price.idx', 'cons.conf.idx',
               'euribor3m', 'nr.employed']

fig, axes = plt.subplots(10, 2, figsize=(16, 24),
                         gridspec_kw={'height_ratios': [4, 1] * 5})


def graficar_pares(columna, slot, fila):
    ax_hist = axes[fila, slot]
    ax_box = axes[fila + 1, slot]

    sns.histplot(ax=ax_hist, data=df, x=columna).set(xlabel=None)
    sns.boxplot(ax=ax_box, data=df, x=columna)

    # Alinear el eje x del histograma con el del boxplot
    ax_hist.sharex(ax_box)


for i, numerica in enumerate(numericas):
    fila = (i // 2) * 2  
    slot = i % 2        
    graficar_pares(numerica, slot, fila)

# Ajustar el layout y mostrar (siempre al final)
plt.tight_layout()
plt.show()
```

- La mayoría de gente cuyos datos se recabaron están entre 30 y 40 años.
- La mayoría de llamadas tuvieron una duración de unos 5 minutos. También es alto el volumen de llamadas inferiores a 1 minuto de duración, lo que indica que muchos clientes cuelgan directamente sin escuchar la promoción del banco.
- Apenas se realizan 2 o 3 contactos a cada cliente durante las campañas.
- A muchos clientes no se les vuelve a contactar desde la última campaña ya que seguramente así lo solicitan ellos mismos. Aún así, se ve como hay valores muy bajos de clientes a los que se vuelve a contactar apenas se realiza la campaña, seguramente porque mostraron interés y estaban abiertos a volver a ser contactados.
- Los clientes que se contactan nunca fueron contratados durante la anterior campaña. Esto se puede deber a que sean clientes nuevos, que no hayan pedido dejar de recibir llamadas comerciales del banco, o que no hace falta contactarles porque ya tienen algún servicio similar contratado.
- Hay una tasa de variabilidad de empleo mayoritariamente positiva.
- Los índices de precio superan muchas veces el 90%, mientras que los de confianza muestran datos preocupantemente bajos.
- La tasa del euribor tiende a ser muy alta en cada cliente.
- El número de empleados siempre es alto, entorno a los 5150.

```python
#primero de todo voy a ir haciendo la factorizacion de las variables tipo string y guardando todo en jsons
import json

# factorizo todas las variables categoricas
for col in categoricas:
    df[col + '_n'] = pd.factorize(df[col])[0]

# json con reglas de conversion
for col in categoricas:
    reglas = {row[col]: row[col + '_n'] for index, row in df[[col + '_n', col]].drop_duplicates().iterrows()}
    with open(f"./{col}_transformation_rules.json", "w") as f:
        json.dump(reglas, f)
```

```python
from numpy._core.defchararray import upper
# Guardar datasets

total_data_CON_outliers = df.copy()
```

Antes ya pude comprobar que no había nulos, así que es otro paso que nos podemos saltar en este proyecto en particular.

Paso a hacer el split en train y test del conjunto de datos para entrenar el modelo. 

```python
from sklearn.model_selection import train_test_split

target = "y_n"
predictoras = [col for col in final_cols if col != target]

X_CON = total_data_CON_outliers[predictoras]
y = total_data_CON_outliers[target]

X_train_CON_outliers, X_test_CON_outliers, y_train, y_test = train_test_split(
    X_CON, y, test_size=0.2, random_state=10
)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler


## NORMALIZACIÓN (cambiar valores para que tengan media de 0 y std de 1)
### CON OUTLIERS
norm_CON_outliers = StandardScaler()
norm_CON_outliers.fit(X_train_CON_outliers)

X_train_CON_outliers_norm = norm_CON_outliers.transform(X_train_CON_outliers)
X_train_CON_outliers_norm = pd.DataFrame(X_train_CON_outliers_norm, index = X_train_CON_outliers.index, columns = predictoras)

X_test_CON_outliers_norm = norm_CON_outliers.transform(X_test_CON_outliers)
X_test_CON_outliers_norm = pd.DataFrame(X_test_CON_outliers_norm, index = X_test_CON_outliers.index, columns = predictoras)


## ESCALADO MIN-MAX (MINMAXIMIZACIÓN) (cambiar valores para que estén comprendidos entre 0 y 1)
### CON OUTLIERS
scaler_CON_outliers = MinMaxScaler()
scaler_CON_outliers.fit(X_train_CON_outliers)

X_train_CON_outliers_scal = scaler_CON_outliers.transform(X_train_CON_outliers)
X_train_CON_outliers_scal = pd.DataFrame(X_train_CON_outliers_scal, index = X_train_CON_outliers.index, columns = predictoras)

X_test_CON_outliers_scal = scaler_CON_outliers.transform(X_test_CON_outliers)
X_test_CON_outliers_scal = pd.DataFrame(X_test_CON_outliers_scal, index = X_test_CON_outliers.index, columns = predictoras)
```

Selección de mejores variables

```python
from sklearn.feature_selection import f_classif, SelectKBest

X_train = X_train_CON_outliers.copy()
X_test = X_test_CON_outliers.copy()

# Con un valor de k = 5 decimos implícitamente que queremos eliminar 2 características del conjunto de datos
selection_model = SelectKBest(f_classif, k = 5)
selection_model.fit(X_train, y_train)

ix = selection_model.get_support()
X_train_sel = pd.DataFrame(selection_model.transform(X_train), columns = X_train.columns.values[ix])
X_test_sel = pd.DataFrame(selection_model.transform(X_test), columns = X_test.columns.values[ix])

X_train_sel.head()
```

Entrenamos el modelo con el conjunto generado anteriormente

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

Con una precisión del 90%, por esta vez no considero necesario hacer una optimización de híper parámetros, y puedo dar por concluido el proyecto.
