```python
# Lista de variables numéricas
numericas = ['age', 'duration', 'campaign', 'pdays', 'previous', 'emp.var.rate', 'cons.price.idx', 'cons.conf.idx',
               'euribor3m', 'nr.employed']

fig, axes = plt.subplots(10, 2, figsize=(16, 24),
                         gridspec_kw={'height_ratios': [4, 1] * 5})


def graficar_pares(columna, slot, fila):
    ax_hist = axes[fila, slot]
    ax_box = axes[fila + 1, slot]

    sns.histplot(ax=ax_hist, data=df, x=columna).set(xlabel=None)
    sns.boxplot(ax=ax_box, data=df, x=columna)

    # Alinear el eje x del histograma con el del boxplot
    ax_hist.sharex(ax_box)


for i, numerica in enumerate(numericas):
    fila = (i // 2) * 2  
    slot = i % 2        
    graficar_pares(numerica, slot, fila)

# Ajustar el layout y mostrar (siempre al final)
plt.tight_layout()
plt.show()
```

- Apenas se realizan 2 o 3 contactos a cada cliente durante las campañas.
-  Prácticamente casi todos los clientes no habían sido contactados antes. 
- La gran mayoría de los clientes (86%) no fue contactada en la campaña anterior.
- Hay una tasa de variabilidad de empleo mayoritariamente positiva.
- Los índices de precio superan muchas veces el 90%, mientras que los de confianza muestran datos preocupantemente bajos.
- La tasa del euribor tiende a ser muy alta en cada cliente.

Antes ya pude comprobar que no había nulos, así que es otro paso que nos podemos saltar en este proyecto en particular.

Paso a hacer el split en train y test del conjunto de datos para entrenar el modelo. 

```python
from sklearn.model_selection import train_test_split

target = "y_n"
predictoras = [col for col in final_cols if col != target] #todas las columnas menos la target

X_CON = total_data_CON_outliers[predictoras]
y = total_data_CON_outliers[target]

X_train_CON_outliers, X_test_CON_outliers, y_train, y_test = train_test_split(
    X_CON, y, test_size=0.2, random_state=10, stratify=y  # stratify para mantener el mismo % de yes en train y test
)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler


## NORMALIZACIÓN (cambiar valores para que tengan media de 0 y std de 1)
### CON OUTLIERS
norm_CON_outliers = StandardScaler()
norm_CON_outliers.fit(X_train_CON_outliers)

X_train_CON_outliers_norm = norm_CON_outliers.transform(X_train_CON_outliers)
X_train_CON_outliers_norm = pd.DataFrame(X_train_CON_outliers_norm, index = X_train_CON_outliers.index, columns = predictoras)

X_test_CON_outliers_norm = norm_CON_outliers.transform(X_test_CON_outliers)
X_test_CON_outliers_norm = pd.DataFrame(X_test_CON_outliers_norm, index = X_test_CON_outliers.index, columns = predictoras)


## ESCALADO MIN-MAX (MINMAXIMIZACIÓN) (cambiar valores para que estén comprendidos entre 0 y 1)
### CON OUTLIERS
scaler_CON_outliers = MinMaxScaler()
scaler_CON_outliers.fit(X_train_CON_outliers)

X_train_CON_outliers_scal = scaler_CON_outliers.transform(X_train_CON_outliers)
X_train_CON_outliers_scal = pd.DataFrame(X_train_CON_outliers_scal, index = X_train_CON_outliers.index, columns = predictoras)

X_test_CON_outliers_scal = scaler_CON_outliers.transform(X_test_CON_outliers)
X_test_CON_outliers_scal = pd.DataFrame(X_test_CON_outliers_scal, index = X_test_CON_outliers.index, columns = predictoras)
```

```python
# GUARDAR TODO 

# PATH = "/data/processed"
PATH = "."

# DATASETS QUE HE IDO ACUMULANDO EN LOS PASOS 4 Y 5
X_train_CON_outliers.to_csv(f"{PATH}/X_train_CON_outliers.csv", index = False)
X_train_CON_outliers_norm.to_csv(f"{PATH}/X_train_CON_outliers_norm.csv", index = False)
X_train_CON_outliers_scal.to_csv(f"{PATH}/X_train_CON_outliers_scal.csv", index = False)


X_test_CON_outliers.to_csv(f"{PATH}/X_test_CON_outliers.csv", index = False)
X_test_CON_outliers_norm.to_csv(f"{PATH}/X_test_CON_outliers_norm.csv", index = False)
X_test_CON_outliers_scal.to_csv(f"{PATH}/X_test_CON_outliers_scal.csv", index = False)


y_train.to_csv(f"{PATH}/y_train.csv", index = False)
y_test.to_csv(f"{PATH}/y_test.csv", index = False)

# SCALERS
import pickle

# MODEL_PATH = "/model"
MODEL_PATH = "."

with open(f"{MODEL_PATH}/norm_CON_outliers.pkl", "wb") as file:
  pickle.dump(norm_CON_outliers, file)

with open(f"{MODEL_PATH}/scaler_CON_outliers.pkl", "wb") as file:
  pickle.dump(scaler_CON_outliers, file)
```

## Modelado

Antes de entrenar hay que tener en cuenta que duration no es una variable muy fiable, por lo que seria mejor prescindir de ella

```python
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score

# funcion para evaluar los modelos
def evaluar(modelo, X_test, y_test):
    y_pred = modelo.predict(X_test)
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    print("AUC:", round(roc_auc_score(y_test, modelo.predict_proba(X_test)[:, 1]), 3))

# datos escalados sin duration (recordemos que no es una variable muy fiable)
X_train_nod = X_train_CON_outliers_norm.drop(columns = 'duration')
X_test_nod = X_test_CON_outliers_norm.drop(columns = 'duration')
```

### Selección de mejores variables

```python
from sklearn.feature_selection import f_classif, SelectKBest
from sklearn.linear_model import LogisticRegression

# Con un valor de k = 5 me quedo con 5 de las 19 variables (ya sin duration)
selection_model = SelectKBest(f_classif, k = 5)
selection_model.fit(X_train_nod, y_train)

ix = selection_model.get_support()
X_train_sel = pd.DataFrame(selection_model.transform(X_train_nod), columns = X_train_nod.columns.values[ix])
X_test_sel = pd.DataFrame(selection_model.transform(X_test_nod), columns = X_test_nod.columns.values[ix])

print(X_train_sel.columns.tolist())

# compruebo si un modelo con solo estas 5 variables mejora al que usa las 19
model_sel = LogisticRegression()
model_sel.fit(X_train_sel, y_train)
evaluar(model_sel, X_test_sel, y_test)
```

Como hay peor rendimiento que con el modelo que usa todas las variables, vamos a continuar con todas en vez de hacer la seleccion.

### Entrenamiento del modelo

Primero con duration y después sin ella. 

```python
#  con duration, saldran unas metricas exageradamente buenas lo mas seguro
model_con_duration = LogisticRegression()
model_con_duration.fit(X_train_CON_outliers_norm, y_train)

evaluar(model_con_duration, X_test_CON_outliers_norm, y_test)
```

```python
# sin duration, saldran peores metricas que con el otro, pero mas realistas
model = LogisticRegression()
model.fit(X_train_nod, y_train)

evaluar(model, X_test_nod, y_test)
```

### Optimización de hiperparámetros

Voy a hacerlo con un grid search. El dicccionario para optimizar será pequeño ya que codespaces no ofrece mucha potencia pero tampoco quiero usar un randomized.

```python
from sklearn.model_selection import GridSearchCV

# diccionario de hiper parametros
param_grid = {"C": [0.01, 0.1, 1, 10],
              "class_weight": [None, "balanced"]}

grid = GridSearchCV(LogisticRegression(max_iter = 1000), param_grid, scoring = "f1", cv = 3, n_jobs = -1)
grid.fit(X_train_nod, y_train)

print("Mejores hiperparametros:", grid.best_params_)
print("F1 en validacion:", round(grid.best_score_, 3))
```

```python
# evaluo el mejor modelo con el test, que no se ha usado en la busqueda
model_opt = grid.best_estimator_
evaluar(model_opt, X_test_nod, y_test)

# guardo el modelo final
with open(f"{MODEL_PATH}/logistic_regression_model.pkl", "wb") as file:
    pickle.dump(model_opt, file)
```
